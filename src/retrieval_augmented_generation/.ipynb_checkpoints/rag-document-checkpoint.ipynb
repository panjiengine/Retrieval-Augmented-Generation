## Buka pdf

```python
import pymupdf # Untuk mengkonversi pdf
from pathlib import Path # untuk path dimana pdf itu berada

PDF_PATH = 'electronics-15-04648.pdf'
# Kode PDF_PATH digunakan untuk membuka file pdf
```

```python
doc = pymupdf.open(PDF_PATH) # Membuka pdf menggunakan pymupdf
```

```python
pages = []

for page_num, page in enumerate(doc, start=1):
    text = page.get_text("text").strip()

    if text:
        pages.append({
            "page": page_num,
            "text": text
        })

print("Jumlah halaman PDF:", len(doc))
print("Halaman yang berisi teks:", len(pages))

if pages:
    print(pages[0]["text"][:1000])
```

```python
def make_chunks(pages, chunk_size=1000, overlap=150):
    chunks = []

    for item in pages:
        text = item["text"]
        start = 0

        while start < len(text):
            end = min(start + chunk_size, len(text))
            chunk = text[start:end].strip()

            if chunk:
                chunks.append({
                    "page": item["page"],
                    "text": chunk
                })

            if end == len(text):
                break

            start = end - overlap

    return chunks


chunks = make_chunks(pages)

print("Jumlah chunks:", len(chunks))
print("\nContoh chunk:")
print(chunks[0]["text"])
print("\nHalaman:", chunks[0]["page"])
```

### Mengubah Chunks menjadi Embedding Granite
Pastikan `llama-server` dengan Model granite sedang berjalan di terminal lain pada port `8080` 
\
Catatan: Port tersebut dinamis dan dapat berubah rubah

```python
import requests # Digunakan untuk memanggil url dari LLama.cpp
import numpy as np

API_URL = "http://127.0.0.1:8080/v1/embeddings" # Dapat di ubah menjadi localhost
```

```python
def embed_texts(texts, batch_size=16):
    all_vectors = []

    for start in range(0, len(texts), batch_size):
        batch = texts[start:start + batch_size]

        response = requests.post(
            API_URL,
            json={"input": batch},
            timeout=180
        )
        response.raise_for_status()

        data = sorted(
            response.json()["data"],
            key=lambda item: item["index"]
        )

        vectors = [
            item["embedding"]
            for item in data
        ]

        all_vectors.extend(vectors)

    return np.asarray(all_vectors, dtype=np.float32)


texts = [chunk["text"] for chunk in chunks]

embeddings = embed_texts(texts)

print("Shape embedding:", embeddings.shape)
```

## Masukkan embedding ke FAISS
Kita gunakan *cosine similarity* melalui inner product pada vektor yang sudah dinormalisasi, sampai disni, jurnal sudah dapat dicari berdasarkan kemiripan semantik.

```python
import faiss

faiss.normalize_L2(embeddings)
dimension = embeddings.shape[1]
index = faiss.IndexFlatIP(dimension)
index.add(embeddings)

print("Jumlah vektor dalam FAISS", index.ntotal)
```

```python
question = "Resources"

query_vector = embed_texts([question])
faiss.normalize_L2(query_vector)

k = min(5, index.ntotal)
scores, ids = index.search(query_vector, k)

for score, idx in zip(scores[0], ids[0]):
    if idx < 0:
        continue

    item = chunks[idx]

    print("=" * 60)
    print(f"Skor kemiripan: {score:.4f}")
    print(f"Halaman PDF: {item['page']}")
    print(item["text"])
    print()
```
