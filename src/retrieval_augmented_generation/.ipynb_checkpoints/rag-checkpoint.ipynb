```python
import requests
import faiss
import numpy as np
```

```python
API_URL = "http://localhost:8080/v1/embeddings"

documents = [
    "Hukum Ohm menyatakan bahwa tegangan sama dengan arus dikali resistansi.",
    "Transistor dapat digunakan sebagai sakelar elektronik.",
    "Dioda memungkinkan arus mengalir terutama ke satu arah.",
    "Linux merupakan sistem operasi yang menggunakan kernel Linux.",
]
```

```python
def embed_texts(texts):
    response = requests.post(
        API_URL,
        json={"input": texts},
        timeout=120,
    )
    response.raise_for_status()

    data = response.json()["data"]
    data = sorted(data, key=lambda item: item["index"])

    return np.asarray(
        [item["embedding"] for item in data],
        dtype=np.float32,
    )
```

```python
doc_vectors = embed_texts(documents)
```

```python
faiss.normalize_L2(doc_vectors)
```

```python
dimension = doc_vectors.shape[1]
index = faiss.IndexFlatIP(dimension)
index.add(doc_vectors)
```

```python
dimension = doc_vectors.shape[1]
index = faiss.IndexFlatIP(dimension)
index.add(doc_vectors)

print("Jumlah dokumen:", index.ntotal)
print("Dimensi embedding:", dimension)
```

```python
question = "Apa itu hukum ohm"

query_vector = embed_texts([question])
faiss.normalize_L2(query_vector)

scores, ids = index.search(query_vector, k=2)

for score, idx in zip(scores[0], ids[0]):
    print(f"Skor: {score:.4f}")
    print(documents[idx])
    print()
```
